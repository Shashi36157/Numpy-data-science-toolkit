```python
# NumPy is used for numerical computation and array operations.
import numpy as np

# Fix the random seed so the generated dataset is reproducible.
# Running this notebook again will generate the same random values.
np.random.seed(42)
```

```python
# Number of students in our simulated dataset.
n = 1000

# Generate unique student IDs from 1001 to 2000.
student_id = np.arange(1001, 1001 + n)

# Generate study hours per day, between 0 and 12.
study_hours = np.random.uniform(0, 12, n)

# Generate attendance percentages, between 40 and 100.
attendance = np.random.uniform(40, 100, n)

# Generate assignment scores, between 0 and 100.
assignment_score = np.random.uniform(0, 100, n)

# Generate exam marks using a formula.
# Study hours, attendance, and assignments influence the marks.
# Random noise makes the data less perfectly predictable.
exam_marks = (
    3 * study_hours
    + 0.25 * attendance
    + 0.3 * assignment_score
    + np.random.normal(0, 10, n)
)

# Keep all exam marks within the valid range of 0 to 100.
exam_marks = np.clip(exam_marks, 0, 100)

# Generate department codes from 1 to 5.
department_code = np.random.randint(1, 6, n)

# Combine all six arrays into one 2D array.
# Each column represents a feature; each row represents a student.
data = np.column_stack((
    student_id,
    study_hours,
    attendance,
    assignment_score,
    exam_marks,
    department_code
))

# Display the dimensions and the first five records.
print("Dataset shape:", data.shape)
print("First five rows:\n", data[:5])
```

```python
# Number of dimensions: our dataset is a 2D array.
print("Dimensions:", data.ndim)

# Number of rows and columns.
print("Shape:", data.shape)

# Total number of individual values: 1000 * 6.
print("Total elements:", data.size)

# Data type used to store the array's values.
print("Data type:", data.dtype)

# Memory used by one element, in bytes.
print("Bytes per element:", data.itemsize)

# Total memory used by the array's elements.
print("Total memory:", data.nbytes, "bytes")
```

```python
# Make a copy so the original dataset remains unchanged.
messy_data = data.copy()

# Introduce 10 missing values into each of the four feature columns.
# Columns 1–4 are study hours, attendance, assignments, and exam marks.
for col in range(1, 5):
    missing_rows = np.random.choice(n, size=10, replace=False)
    messy_data[missing_rows, col] = np.nan

# Insert four deliberately invalid values for practice.
messy_data[0, 1] = 30       # Invalid study hours: greater than 12.
messy_data[1, 2] = 150      # Invalid attendance: greater than 100%.
messy_data[2, 3] = -10      # Invalid assignment score: below 0.
messy_data[3, 4] = 150      # Invalid exam marks: greater than 100.

# Count missing values in each column.
# np.isnan returns True for NaN; sum counts those True values.
print("Missing values by column:", np.isnan(messy_data).sum(axis=0))

# Display a few records to inspect the data.
print("First five messy rows:\n", messy_data[:5])
```

```python
# Column 1: study hours must be between 0 and 12.
invalid_study = (
    (messy_data[:, 1] < 0) |
    (messy_data[:, 1] > 12)
)

# Column 2: attendance must be between 0 and 100.
invalid_attendance = (
    (messy_data[:, 2] < 0) |
    (messy_data[:, 2] > 100)
)

# Column 3: assignment scores must be between 0 and 100.
invalid_assignment = (
    (messy_data[:, 3] < 0) |
    (messy_data[:, 3] > 100)
)

# Column 4: exam marks must be between 0 and 100.
invalid_exam = (
    (messy_data[:, 4] < 0) |
    (messy_data[:, 4] > 100)
)

# Print the number of invalid values in each feature.
print("Invalid study hours:", invalid_study.sum())
print("Invalid attendance:", invalid_attendance.sum())
print("Invalid assignment scores:", invalid_assignment.sum())
print("Invalid exam marks:", invalid_exam.sum())

# Display the IDs of students with invalid study hours.
print("Students with invalid study hours:")
print(messy_data[invalid_study, 0])
```

```python
# Feature names correspond to columns 1 through 4.
feature_names = [
    "Study Hours",
    "Attendance",
    "Assignment Score",
    "Exam Marks"
]

# Analyze each feature separately.
for i, name in enumerate(feature_names, start=1):
    column = messy_data[:, i]

    print(f"\n{name}")

    # Calculate statistics while ignoring missing values.
    print("Mean:", np.nanmean(column))
    print("Median:", np.nanmedian(column))
    print("Minimum:", np.nanmin(column))
    print("Maximum:", np.nanmax(column))
    print("Standard deviation:", np.nanstd(column))
```

```python
# Create a separate copy for cleaning.
cleaned_data = messy_data.copy()

# Replace invalid values with NaN rather than guessing their true values.
cleaned_data[invalid_study, 1] = np.nan
cleaned_data[invalid_attendance, 2] = np.nan
cleaned_data[invalid_assignment, 3] = np.nan
cleaned_data[invalid_exam, 4] = np.nan

# Verify the number of missing values after invalid values are handled.
print("Missing values by column:")
print(np.isnan(cleaned_data).sum(axis=0))
```

```python
# Process the four numerical feature columns.
for col in range(1, 5):

    # Calculate the median while ignoring NaN values.
    median_value = np.nanmedian(cleaned_data[:, col])

    # Find missing positions in the current column.
    missing_mask = np.isnan(cleaned_data[:, col])

    # Replace missing values with the column median.
    cleaned_data[missing_mask, col] = median_value

    # Report how many values were filled.
    print(
        feature_names[col - 1],
        "| Median used:", round(median_value, 2),
        "| Values filled:", missing_mask.sum()
    )

# Check that no missing values remain.
print("\nTotal missing values:", np.isnan(cleaned_data).sum())
```

```python
# Add one duplicate record so we can practice duplicate removal.
duplicate_test = np.vstack((
    cleaned_data,
    cleaned_data[10]
))

print("Rows before removal:", duplicate_test.shape[0])

# Extract all student IDs.
student_ids = duplicate_test[:, 0]

# Find each unique ID and the index of its first occurrence.
unique_ids, first_indices = np.unique(
    student_ids,
    return_index=True
)

# Restore the original row order.
first_indices.sort()

# Keep only the first record for each student ID.
cleaned_data = duplicate_test[first_indices]

print("Rows after removal:", cleaned_data.shape[0])
print("Unique student IDs:", np.unique(cleaned_data[:, 0]).size)
```

```python
# Examine each feature for potential outliers.
for col, name in enumerate(feature_names, start=1):

    values = cleaned_data[:, col]

    # Calculate the first and third quartiles.
    q1 = np.percentile(values, 25)
    q3 = np.percentile(values, 75)

    # Interquartile range measures the spread of the middle 50%.
    iqr = q3 - q1

    # Define the lower and upper outlier thresholds.
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    # Flag values outside these thresholds.
    outlier_mask = (
        (values < lower_bound) |
        (values > upper_bound)
    )

    print(f"\n{name}")
    print("Lower bound:", round(lower_bound, 2))
    print("Upper bound:", round(upper_bound, 2))
    print("Potential outliers:", outlier_mask.sum())
```

```python
# Extract each feature from the cleaned dataset.
# These variables will be used in the remaining analysis.
study = cleaned_data[:, 1]
attendance = cleaned_data[:, 2]
assignment = cleaned_data[:, 3]
exam = cleaned_data[:, 4]
department = cleaned_data[:, 5]

# Calculate overall descriptive statistics.
print("Average study hours:", round(np.mean(study), 2))
print("Average attendance:", round(np.mean(attendance), 2))
print("Average assignment score:", round(np.mean(assignment), 2))
print("Average exam marks:", round(np.mean(exam), 2))

print("Highest exam marks:", round(np.max(exam), 2))
print("Lowest exam marks:", round(np.min(exam), 2))
print("Median exam marks:", round(np.median(exam), 2))
print("Exam mark standard deviation:", round(np.std(exam), 2))
```

```python
# np.argsort returns the indices that would sort exam marks in ascending order.
# [-10:] selects the indices of the 10 largest marks.
# [::-1] reverses their order so the highest mark appears first.
top_indices = np.argsort(exam)[-10:][::-1]

# Use these indices to retrieve the complete student records.
top_students = cleaned_data[top_indices]

print("Top 10 students")
print("Student ID | Exam Marks")

for row in top_students:
    # Convert the ID to an integer for cleaner display.
    # Round exam marks to two decimal places.
    print(int(row[0]), "|", round(row[4], 2))
```

```python
# Find all distinct department codes.
departments = np.unique(department)

for dept in departments:

    # Select students belonging to this department.
    mask = department == dept

    # Calculate average exam marks for the selected students.
    avg_exam = np.mean(exam[mask])

    # Count the students in this department.
    student_count = np.sum(mask)

    print(
        f"Department {int(dept)}: "
        f"Students = {student_count}, "
        f"Average exam marks = {avg_exam:.2f}"
    )
```

```python
# Define a simple rule for identifying students who may need support:
# exam marks below 40 AND attendance below 75%.
needs_support = (exam < 40) & (attendance < 75)

# Extract all matching student records.
support_students = cleaned_data[needs_support]

print("Students needing support:", support_students.shape[0])
print("Student IDs:")
print(support_students[:, 0])
```

```python
# Select the four numerical input features.
# Shape: (1000, 4)
X = cleaned_data[:, 1:5]

# Calculate the minimum and maximum of each column.
# axis=0 means we calculate these values column by column.
minimum = np.min(X, axis=0)
maximum = np.max(X, axis=0)

# Calculate the range of each feature.
feature_range = maximum - minimum

# Avoid division by zero if a feature has identical values throughout.
safe_range = np.where(feature_range == 0, 1, feature_range)

# Apply min-max scaling using broadcasting.
X_normalized = (X - minimum) / safe_range

print("First 5 normalized rows:")
print(np.round(X_normalized[:5], 3))

# Verify the minimum and maximum of each normalized feature.
print("Feature minima:", np.min(X_normalized, axis=0))
print("Feature maxima:", np.max(X_normalized, axis=0))
```

```python
# Calculate Pearson correlations between the four numerical features.
# rowvar=False tells NumPy that columns represent variables.
correlation_matrix = np.corrcoef(X, rowvar=False)

# Print a rounded version for easier interpretation.
print("Feature order:")
print(feature_names)

print("\nCorrelation matrix:")
print(np.round(correlation_matrix, 2))
```

```python
def analyze_array(arr):
    """
    Calculate basic descriptive statistics for a numerical array.

    Parameters:
        arr: A non-empty array containing finite numerical values.

    Returns:
        A dictionary containing the shape, mean, median,
        standard deviation, minimum, and maximum.
    """

    # Convert the input into a numerical NumPy array.
    arr = np.asarray(arr, dtype=float)

    # Reject empty input.
    if arr.size == 0:
        raise ValueError("Array cannot be empty")

    # Reject NaN and infinite values.
    if not np.all(np.isfinite(arr)):
        raise ValueError("Array must contain only finite values")

    # Return the calculated statistics.
    return {
        "shape": arr.shape,
        "mean": np.mean(arr),
        "median": np.median(arr),
        "std": np.std(arr),
        "min": np.min(arr),
        "max": np.max(arr)
    }


# Analyze the exam marks and display the returned dictionary.
exam_report = analyze_array(exam)

for key, value in exam_report.items():
    print(f"{key}: {value}")
```

```python
# Count students satisfying our example support rule.
support_count = np.sum((exam < 40) & (attendance < 75))

# Print a concise report summarizing our work.
print("=" * 45)
print("NUMPY DATA SCIENCE TOOLKIT — FINAL REPORT")
print("=" * 45)

print("Total students:", cleaned_data.shape[0])
print("Total columns:", cleaned_data.shape[1])
print("Remaining missing values:", np.isnan(cleaned_data).sum())
print("Unique student IDs:", np.unique(cleaned_data[:, 0]).size)

print("Average exam marks:", round(np.mean(exam), 2))
print("Highest exam marks:", round(np.max(exam), 2))
print("Lowest exam marks:", round(np.min(exam), 2))

print("Students flagged for support:", support_count)

# Validate the key properties of the final dataset.
assert cleaned_data.shape == (1000, 6)
assert np.isnan(cleaned_data).sum() == 0
assert np.unique(cleaned_data[:, 0]).size == 1000

print("\nAll final validation checks passed!")
print("=" * 45)
```
